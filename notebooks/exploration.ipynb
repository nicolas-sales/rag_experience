### Observation — tableau multi-page

Le tableau "Liste des États contractants et signataires de la Convention"
s'étend sur 3 pages physiques consécutives du PDF :

- index PyMuPDF 34
- index PyMuPDF 35
- index PyMuPDF 36

`page.find_tables()` ne reconstruit pas correctement le tableau complet.
Sur ce document, les lignes avec fond coloré sont principalement détectées
comme des tables indépendantes de 1 ligne × 4 colonnes, tandis que plusieurs
lignes sans fond coloré ne sont pas détectées comme tables.

En revanche, `page.get_text("dict")` récupère bien le texte des lignes
manquantes ainsi que leurs bounding boxes. Les X pour les colonnes et les y pour les lignes.

Conclusion : le contenu est disponible dans le PDF, mais une reconstruction
layout-aware est nécessaire pour obtenir le tableau logique complet.

```python
page = doc[34]
page_dict = page.get_text("dict")

target_y = 298.13
tolerance = 0.5

same_row_spans = []

for block in page_dict["blocks"]:
    if block["type"] != 0:
        continue

    for line in block["lines"]:
        for span in line["spans"]:
            x0, y0, x1, y1 = span["bbox"]

            if abs(y0 - target_y) <= tolerance: # Si la distance antre les 2 valeurs inférieures à tolérance
                same_row_spans.append({ # Appartiennent à la même ligne
                    "text": span["text"],
                    "x0": x0,
                    "y0": y0,
                    "bbox": span["bbox"],
                })

same_row_spans = sorted(
    same_row_spans,
    key=lambda span: span["x0"]  # on trie de gauche à droite selon leur x0, l'ordre spatial
)

for span in same_row_spans:
    print(
        f"x={span['x0']:.1f} | "
        f"y={span['y0']:.1f} | "
        f"text={span['text']!r}"
    )
```

```python
page = doc[34]
page_dict = page.get_text("dict")

left_spans = []

for block in page_dict["blocks"]:
    if block["type"] != 0:
        continue

    for line in block["lines"]:
        for span in line["spans"]:
            x0, y0, x1, y1 = span["bbox"]

            if x0 < 550:
                left_spans.append({
                    "text": span["text"].strip(), # etire les espaces inutiles au début et à la fin du texte
                    "x0": x0,
                    "y0": y0,
                    "x1": x1,
                    "y1": y1,
                    "bbox": span["bbox"],
                })
```

```python
table_spans = [
    span
    for span in left_spans
    if 290 <= span["y0"] <= 700 # Garde les spans du panneau gauche dont la position verticale se situe entre 290 et 700
]

print("Nombre de spans :", len(table_spans))
```

```python
rows = []
tolerance = 0.5

for span in sorted(table_spans, key=lambda s: s["y0"]): # on trie tous les spans du haut vers le bas

    matching_row = None

    for row in rows:
        if abs(span["y0"] - row["y"]) <= tolerance: # y a t il une ligne ayant approximativement le même y?
            matching_row = row
            break # si oui, stoppe la recherche

    if matching_row is None: # Si aucune ligne ne correspond
        rows.append({        # On crée une nouvelle ligne
            "y": span["y0"],
            "spans": [span],
        })
    else:
        matching_row["spans"].append(span) # on ajoute le span à la ligne existante
```

```python
for row in rows:
    row["spans"] = sorted(
        row["spans"],
        key=lambda s: s["x0"]
    )

    texts = [
        span["text"]
        for span in row["spans"]
        if span["text"]
    ]

    print(f"y={row['y']:.1f} -> {texts}")
```

```python
def get_column_index(x0):
    for column_index, (x_min, x_max) in enumerate(column_ranges):
        if x_min <= x0 < x_max:
            return column_index

    return None
```

# Mesure des écarts entre les x

```python
x_gaps = []

for i in range(len(x_positions_sorted) -1): # -1 car dans la boucle on utilise x_positions_sorted[i + 1] et si on allait jusqu’au dernier élément, il n’existerait plus d’élément suivant
    gap = x_positions_sorted[i+1] - x_positions_sorted[i]
    x_gaps.append(gap)

print(x_gaps[:20])
```

### Fonction avec tolérance

```python
def group_close_positions(positions, tolerance):
    """
    positions : liste de x triés
    tolerance : distance maximale qu'on accepte entre deux positions voisines pour considérer qu'elles appartiennt au même groupe
    """
    groups = [[positions[0]]] # premier élément de la liste et on veut une liste de groupe

    for x in positions[1:]: # commence au 2ème élément de la liste
        if x - groups[-1][-1] <= tolerance: # prend le dernier groupe et le dernier élément du dernier groupe
            groups[-1].append(x) # ajoute x au groupe actuel

        else:
            groups.append([x]) # sinon crée un nouveau groupe

    return groups
```

```python
groups = group_close_positions(x_positions_sorted, tolerance=1)

for group in groups:
    print(len(group), round(sum(group) / len(group), 1))
```

La fonction a découvert 4 groupes de positions, donc les 4 colonnes du panneau.

Pour les deux derniers, la longueur est de 19 car pour Bélize, les deux dernières colonnes sont vides.

### Définition de la tolérance

```python
# On retire les gaps nuls pour éviter la division par 0

non_zero_gaps = [ gap for gap in largest_gaps if gap > 0]
```

```python
# Comparaison de chaque grand gap au suivant

gap_ratios = [] # liste qui va contenir des rapports, et non plus des distances

for i in range(len(non_zero_gaps) - 1):
    ratio = non_zero_gaps[i] / non_zero_gaps[i + 1]
    gap_ratios.append(ratio)

print(gap_ratios[:10])
```

Deux ruptures, 408 et 27

```python
# Plus grand ratio et sa position

max_ratio = max(gap_ratios)
rupture_index = gap_ratios.index(max_ratio)

print("Plus grans ration :", max_ratio)
print("Index de rupture :", rupture_index)
```

```python
large_gap = non_zero_gaps[rupture_index]
small_gap = non_zero_gaps[rupture_index + 1]

auto_tolerance = (large_gap + small_gap) / 2

print("Large gap :", large_gap)
print("Small gap :", small_gap)
print("Tolérance automatique :", auto_tolerance)
```

Deux positions x séparées de moiuns de 50 points seont considérés comme appartenant au même groupe, au delà on crée un nouveau groupe

```python
def group_close_positions(positions, tolerance):
    positions = sorted(positions)
    groups = [[positions[0]]]

    for x in positions[1:]:
        if x - groups[-1][-1] <= tolerance:
            groups[-1].append(x)
        else:
            groups.append([x])

    return groups

group_close_positions(x_position, tolerance=auto_tolerance)
```

```python
for group in groups:
    print(len(group), round(sum(group) / len(group), 1))
```
